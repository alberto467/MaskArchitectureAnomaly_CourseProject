```python
import wandb
api = wandb.Api()
run = api.run("/alberto467-politecnico-di-torino/coco_model_cityscape_finetune/runs/djix312w")
```

```python
import math

EVERY_STEP = 100

steps = {}

for history in run.history():
    def filter_key(k):
        if "_block_" in k:
            return False
        for allow_key in ["metrics", "trainer", "losses", "epoch"]:
            if k.startswith(allow_key):
                return True
    history = { k: o for k, o in history.items() if filter_key(k) }
    
    global_step = history["trainer/global_step"]
    rounded_key = math.ceil(global_step/EVERY_STEP)*EVERY_STEP
    steps[rounded_key] = { **steps.get(rounded_key, {}), **history }

print("\n".join([ f"{o}" for k, o in steps.items() if k != 0 ]))
```
